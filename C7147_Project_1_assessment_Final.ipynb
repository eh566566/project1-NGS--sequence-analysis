## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence: str, reference: str) -> dict:
    """Checks sequences are suitable: right length, only DNA bases, transforms lowercase sequences to uppercase if needed.
    Returns a dictionary with the (transformed) sequence, its validity and it's errors."""

    #initiates a list in the dictionary
    errors = []
    # transforms lowercase sequence into uppercase() if needed - sequence input
    sequence = sequence.upper()

    # checks length - for valid input, errors
    if len(sequence) != len(reference):
          errors.append("wrong_length")

    # checks if special characters are present - for valid input, errors
    if not all(base in "ACTG" for base in sequence):
          errors.append("invalid_characters")

    # validity status
    if len(errors) > 0:
        valid = False
    else:
        valid = True

    # returns output in a dictionary for each individual read
    return {"sequence": sequence, "valid": valid, "errors": errors}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence:str, reference:str) -> list:
    """Find mutations and their position between the reference and the sequence, giving the differing bases at each position"""

    invalid_reads = ["sequence not valid"]
    valid_reads = validate_read(sequence, reference)
    if valid_reads["valid"] == False:
        return invalid_reads
    else:
    # generates an empty list
        mutations = []
        # for each position, base 1 of the sequence is compared to base 1 of the reference and appended to mutations if differing
        for position, (base_1, base_2) in enumerate(zip(sequence, reference)):
            if base_1 != base_2:
                mutations.append({"position": position + 1, "reference_base": base_2, "read_base": base_1})
        # hands output back
        return mutations
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads:list, reference:str)->list:
    """receives input from validate_read() and find_mutations() and generates a list with one dictionary per read, returning all information
    combined from both functions so far, including read, sample, group, sequence, status, errors and mutations if sequence is valid. """

    # generates an empty list for output
    analysed_reads = []

    # checks validity for each read
    for read in reads:
        validity = validate_read(read["sequence"], reference)

        # validity is stated under new key status in dictionary, mutations are only added if read is valid
        # if statement prevents later wrong calculations
        if validity["valid"] == True:
            status = "valid"
            mutations = find_mutations(validity["sequence"], reference)
        else:
            status = "invalid"
            mutations = []

        # for each read a dictionary is generated
        analysis = {"read_id": read["read_id"],
                    "sample": read["sample"],
                    "group": read["group"],
                    "sequence": validity["sequence"],
                    "status": status,
                    "errors": validity["errors"],
                    "mutations": mutations
                        }
        # appends start list with dictionary analysis for each read
        analysed_reads.append(analysis)

    #hands output back
    return analysed_reads
````

## Step 4 — Summarise mismatch patterns between samples

## Function: `summarise_samples(analysed_reads)`

### 👍 You can assume:

````python
# turns output into something that can be passed into a new function
analysed_reads = analyse_reads(reads, reference)

def summarise_samples(analysed_reads:list)->dict:
    """yields input from analyse_reads() and summarises reads by samples  - giving group, number of valid and invalid reads,
     mismatch reads(number of reads containing mismatches), mismatch percentage(reads with mismatches vs valid reads),
     mismatch events(number of mismatches in sample), as well as base substitution counts, position counts and top positions
     (positions with highest mismatches)"""

    # generates a new dictionary
    samples = {}

    # adds all the different samples to the new dictionary as keys and introduces another dictionary
    # with items as values within the outer dictionary
    for read in analysed_reads:
        key = read["sample"]
        if key not in samples:
           samples[key] = {
               "group": read["group"],
               "valid reads": 0,
               "invalid reads": 0,
               "mismatch reads": 0,
               "mismatch percentage": 0,
               "mismatch events": 0,
               "substitution counts":{},
               "position counts":{},
               "top positions": []
           }

        # updates valid and invalid reads in accumulator pattern:
        if  read["status"] == "valid":
            samples[key]["valid reads"] += 1
        else:
            samples[key]["invalid reads"] += 1

        # updates mismatch reads, counts the number of reads with mutations irrelevant of whether they have one or multiple mutation
        if len(read["mutations"]) > 0:
            samples[key]["mismatch reads"] += 1

        # mismatch percentage, the percentage of mismatch reads in comparison to valid reads. This only occurs if valid reads > 0
        if samples[key]["valid reads"] > 0:
           mismatch_percentage = round((samples[key]["mismatch reads"]/(samples[key]["valid reads"]))*100, 1)
        else:
          mismatch_percentage = None

        # assigns and updates mismatch percentage value to the above dictionary
        samples[key]["mismatch percentage"] = mismatch_percentage

        #mismatch events, counts the total number of mutations to the above dictionary, counting one if only one is present in
        #read or multiple if there are multiple in a read
        samples[key]["mismatch events"] += len(read["mutations"])

        # new dictionary for subsitutions within samples dictionary
        substitution_counts = samples[key]["substitution counts"]

        # for loop, tells programme to access mutations list in analysed reads
        for mutation in read["mutations"]:
            substitution = mutation["reference_base"] + ">" + mutation["read_base"]

            # if/else tells to update substitution dictionary if substition is new, else adding a count
            if substitution not in substitution_counts:
                substitution_counts[substitution] = 1
            else:
                substitution_counts[substitution] += 1

        # position counts - the same structure as substitution counts - generates a new dictionary, for loop tells programme
        # to access mutation list within analysed reads, if/else updates position counts within each sample for each sample
        position_counts = samples[key]["position counts"]
        for mutation in read["mutations"]:
            position = mutation["position"]

            if position not in position_counts:
                position_counts[position] = 1
            else:
                position_counts[position] += 1


        # sorts position counts by the positions with the greatest amount of mismatches and appends the top 3 or less
        # if there are less present to top_positions
        top_positions = []
        position_counts_sorted = sorted(position_counts, key = position_counts.get, reverse = True)
        for position in position_counts_sorted:
            if len(top_positions) < 3:
                top_positions.append(position)
        samples[key]["top positions"] = top_positions





    # returns the output instead of printing it, more robust
    return samples

# prints out results in a table: summarised reads
summarised_samples = summarise_samples(analysed_reads)
# table title
print("Table of summarised reads:")

# prints header
print(f'{"sample":<30}{"valid reads":<30}{"invalid reads":<25}{"mismatch reads percentage(%)":<20}{"mismatch events":>30}{"top 3 positions(mismatches)":>30}')
print("-"*180)

# prints rows beneath for each column
for key in summarised_samples:
    print(f'{key:<30}{summarised_samples[key]["valid reads"]:<30}{summarised_samples[key]["invalid reads"]:<25}{summarised_samples[key]["mismatch percentage"]:<20}{summarised_samples[key]["mismatch events"]:>30}{str(summarised_samples[key]["top positions"]):>30}')

# prints substitution table title and header of substitution table
print("Substitution mutations:")
samples = list(summarised_samples.keys())
print(f'{"              ":<40}' + ''.join(f'{sample:<20}' for sample in samples))
print("-" * (35 + 25 * len(samples)))


# prints rows with the substitution occurences and the percentage of substitution occurences
for sample in summarised_samples:
    for substitution, counts in summarised_samples[sample]["substitution counts"].items():
        print(f'{"number of substitution occurences" :<40}{substitution} {counts}')
        percentage_substitution_occurence = round((counts/summarised_samples[sample]["valid reads"])*100, 1)
        print(f'{"percentage substitution occurence":<40}{percentage_substitution_occurence}{"%"}')



````

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python

```

Each list must contain **one percentage per biological sample**.

The supplied function returns:


It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````
